```python
# Recommended for demos: fast & economical
model_name = "gemini-3.6-flash"

llm = ChatGoogleGenerativeAI(
    model=model_name,
    convert_system_message_to_human=True,  # smoother behavior with system prompts
    temperature=0.7, # to enconrage a bit of creativity in responses 
)
print(f" Ready with model: {model_name}")
```

```python
from langchain_core.messages import trim_messages
from langchain_community.chat_message_histories import ChatMessageHistory

# 1. Initialize message history
chat_history = ChatMessageHistory()

# 2. Define message trimmer for the last 2 turns (4 messages)
trimmer = trim_messages(
    max_tokens=4,
    strategy="last",
    token_counter=len,  # Counts message objects directly
    start_on="human",
    include_system=True,
)
```

```python
# Prompt template
prompt = PromptTemplate(
    template="Chat History: {chat_history}\n\nHuman: {input}\nAI:",
    input_variables=["input"],
)
```

```python
base_chain = prompt | llm | StrOutputParser()
```

```python
# Store conversation history per session
store = {}

def get_session_history(session_id: str):
    if session_id not in store:
        store[session_id] = ChatMessageHistory()
    return store[session_id]
```

```python
chat_chain = RunnableWithMessageHistory(
    base_chain,
    get_session_history,
    input_messages_key="input",
    history_messages_key="chat_history",
)
```

```python
# Step 1: Name input
name = input("Enter your name: ")

chat_chain.invoke(
    {"input": f"My name is {name}. Please remember it and greet me."},
    config={"configurable": {"session_id": "user_session_1"}}
)

# Step 2: Topic 1
#chat_chain.invoke({"input": "Tell me something interesting about Mars."})
chat_chain.invoke(
    {"input": "Tell me something interesting about Mars."},
    config={"configurable": {"session_id": "user_session_1"}}
)


# # Step 3: Topic 2
chat_chain.invoke(
    {"input": "Now explain how plants make food."},
    config={"configurable": {"session_id": "user_session_1"}}


)


# # Step 4: Topic 3
chat_chain.invoke(
    {"input": "Now explain Where do we apply calculus in Data Science?."},
    config={"configurable": {"session_id": "user_session_1"}}
    )


# # Step 4: Ask if it still remembers the name (it won’t!)
chat_chain.invoke(
    {"input": "Do you remember my name?"},
    config={"configurable": {"session_id": "user_session_1"}}

    )



 # 

```

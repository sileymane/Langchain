```python
import os


# LangChain + Gemini
import langchain_core
from langchain_google_genai import ChatGoogleGenerativeAI # to work with Geminiai
from langchain_core.prompts import PromptTemplate # TO FORMAT the PROMT 
from langchain_core.output_parsers import StrOutputParser # to extract plain text from responses
from langchain_core.runnables import RunnablePassthrough # chaining steps 
```

```python
# Recommended for demos: fast & economical
model_name = "gemini-3.6-flash"

llm = ChatGoogleGenerativeAI(
    model=model_name,
    convert_system_message_to_human=True,  # smoother behavior with system prompts
    temperature=0.7, # to enconrage a bit of creativity in responses 
)
print(f" Ready with model: {model_name}")
```

```python
#  Prompt to get user's name
name_prompt = PromptTemplate.from_template("My name is {name}. Please remember it and greet me.")
```

```python
#  Prompt to ask if it remembers the name
followup_prompt = PromptTemplate.from_template("Do you remember my name?")
```

```python
# 3. Chain them together using LCEL
# Modern LCEL syntax (Recommended)
name_chain = name_prompt | llm | StrOutputParser()
followup_chain = followup_prompt | llm | StrOutputParser()
```

```python
# #  Run both chains

name = input("Enter your name: ")
greet = name_chain.invoke({"name": name})
print("\n Greet:\n", greet)

# 3. Pass the greeting context into the follow-up chain
print("\n Asking if it remembers your name...")
response = followup_chain.invoke({"greet": greet})

print("\n Response:\n", response)
```
